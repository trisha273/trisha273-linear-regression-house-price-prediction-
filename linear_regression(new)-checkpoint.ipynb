```python
#packages

import numpy as np
import matplotlib.pyplot as plt
```

```python
#house size(in sq.feet)
x_train = np.array([
     720,  980, 1150, 1320, 1080,
    1450, 1210, 1680, 1540, 1390,
    1760, 1120, 1850, 1280, 1510,
    1630, 1190, 1420, 1970, 1340,
    1570, 1260, 1720, 1480, 1100,
    1810, 1360, 1590, 1230, 1900,
    1050, 1440, 1660, 1290, 1750,
    1520, 1180, 1870, 1380, 1610,
     900, 2000, 1470, 1300, 1550,
    2150, 2450, 820, 2700, 1850
], dtype=float)

#house price(in 1000$s)
y_train = np.array([
    145, 178, 205, 235, 190,
    260, 215, 275, 245, 225,
    285, 195, 310, 205, 250,
    270, 220, 245, 330, 218,
    265, 210, 295, 235, 180,
    305, 230, 280, 215, 315,
    170, 255, 285, 225, 290,
    248, 200, 300, 240, 275,
    155, 340, 250, 220, 260,
    365, 395, 135, 430, 305
], dtype=float)
```

```python
#potting the datas with actual values

plt.scatter(x_train,y_train,marker="x",color="red")
plt.title("House size vs House Price")
plt.xlabel("House size(in sq.feet)")
plt.ylabel("House price(in 1000$s)")
plt.show()
```

```python
#calculate cost
def compute_cost(x,y,w,b):
    m=x.shape[0]
    total_cost=0

    for i in range(m):
        f_wb=w*x[i]+b
        cost=(f_wb-y[i])**2
        total_cost+=cost

    total_cost/=(2*m)

    return total_cost
```

```python
#calculate gradient
def compute_gradient(x,y,w,b):
    m=x.shape[0]
    dj_dw=0
    dj_db=0

    for i in range(m):
        f_wb=w*x[i]+b
        
        dj_dw_i=(f_wb-y[i])*x[i]
        dj_db_i=(f_wb-y[i])

        dj_dw+=dj_dw_i
        dj_db+=dj_db_i

    dj_dw/=m
    dj_db/=m

    return dj_dw,dj_db
        
```

```python
#calculate gradient descent
def gradient_descent(x,y,w,b,alpha,iters):
    cost_history=[]

    for i in range(iters):
        dj_dw,dj_db=compute_gradient(x,y,w,b)

        w=w-alpha*dj_dw
        b=b-alpha*dj_db

        cost=compute_cost(x,y,w,b)
        cost_history.append(cost)

    return w,b,cost_history
```

```python
#initialize w and b
w=0
b=0

#give appropriate value of alpha and iterations
alpha=0.03
iters=10000
```

```python
#Feature Scaling applied on x(too large values of x)
x_scaled=x_train/1000

w,b,cost_history=gradient_descent(x_scaled,y_train,w,b,alpha,iters)

```

```python
#NOT COMPULSORY(only to check which value of alpha has min. cost)

w_init=0
b_init=0

alphas = [0.015, 0.02, 0.025, 0.03, 0.035]

for alpha in alphas:

    w, b, cost_history = gradient_descent(
        x_scaled,
        y_train,
        w_init,
        b_init,
        alpha,
        iters
    )

    print("alpha =", alpha)
    print("Final cost =", cost_history[-1])
```

```python
#to see values of w,b and cost
print("Final w =", w)
print("Final b =", b)

print("Initial cost =", cost_history[0])
print("Final cost =", cost_history[-1])
```

```python
#graph of cost vs iterations
plt.plot(cost_history)
plt.title("Cost vs Iterations")
plt.xlabel("Iterations")
plt.ylabel("cost")
plt.show()
```

```python
#final graph of actual and predicted values
y_pred=w*x_scaled+b

plt.scatter(x_train,y_train,marker="x",color="red",label="Training data")
plt.plot(x_train,y_pred,c="blue",label="Regression line")

plt.title("House size vs House Price")
plt.xlabel("House size(in sq.feet)")
plt.ylabel("House price(in 1000$s)")

plt.legend()
plt.show()
```
